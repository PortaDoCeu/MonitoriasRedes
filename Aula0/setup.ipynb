# Configuracao inicial do ambiente TypeScript

Este notebook apresenta os comandos basicos para criar um projeto com TypeScript e executar arquivos `.ts` de forma simples.

Todos os comandos mostrados neste roteiro devem ser executados no terminal do sistema, como o **Prompt de Comando** ou o terminal integrado do **VS Code**.
Caso use o terminal integrado do vscode lembre-se de confirmar que está usando o cmd, pois algumas permissões bloqueam comandos como esse de rodar no powershell



## 0. Instalando o Node.js

Antes de executar os comandos deste roteiro, e necessario instalar o **Node.js**, pois ele inclui o `node` e o `npm`, que serao usados durante toda a configuracao do projeto.

A instalacao pode ser feita pelo site oficial do Node.js.

Depois da instalacao, e recomendavel verificar se tudo foi instalado corretamente com os comandos:

```bash
node -v
npm -v
```

Esses comandos tambem ajudam a confirmar se o Node.js foi adicionado corretamente ao `PATH` do sistema.

Se o terminal nao reconhecer `node` ou `npm`, mesmo apos a instalacao, algumas acoes validas sao:

- fechar e abrir novamente o VS Code;
- abrir um novo terminal;
- reiniciar o computador, para garantir a atualizacao das variaveis de ambiente.

Se, mesmo assim, o comando `node -v` nao funcionar, e necessario verificar se o Node.js foi instalado corretamente e se ele foi incluido no `PATH`.





## 1. Criando o projeto

O comando abaixo cria o arquivo `package.json`, que guarda as informacoes principais do projeto e suas dependencias.

```bash
npm init -y
```


## 2. Instalando o TypeScript

Este comando instala o compilador TypeScript e os tipos do Node.js como dependencias de desenvolvimento.

- `typescript`: permite escrever e compilar arquivos `.ts`.
- `@types/node`: adiciona tipagens para recursos do Node.js, como `console`, `process` e modulos nativos.

```bash
npm install typescript @types/node --save-dev
```


## 3. Gerando o arquivo de configuracao

O comando a seguir cria o arquivo `tsconfig.json`, utilizado para definir as regras de compilacao do TypeScript.

```bash
npx tsc --init
```


## 4. Instalando o tsx

O `tsx` permite executar arquivos TypeScript diretamente, sem a necessidade de compilar manualmente para JavaScript antes.

```bash
npm install tsx --save-dev
```


## 5. Adicionando scripts no `package.json`

O bloco abaixo nao deve ser executado no terminal. Ele deve ser inserido dentro da chave `"scripts"` do arquivo `package.json`.

```json
"scripts": {
  "start": "tsx src/index.ts",
  "dev": "tsx watch src/index.ts",
  "build": "tsc"
}
```

Significado de cada script:

- `start`: executa o arquivo principal uma vez.
- `dev`: executa o projeto em modo de desenvolvimento, atualizando automaticamente quando houver alteracoes.
- `build`: compila os arquivos TypeScript para JavaScript.


## 6. Executando um arquivo TypeScript

Para executar diretamente um arquivo `.ts`, utilize:

```bash
npx tsx src/nome.ts
```

Se os scripts ja tiverem sido adicionados ao `package.json`, tambem e possivel usar:

```bash
npm run start
npm run dev
npm run build
```
