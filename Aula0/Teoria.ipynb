# Introdução ao Sniffer Industrial

Um **sniffer industrial** é uma ferramenta usada para capturar e analisar pacotes de rede em ambientes de automação. Ele permite observar como dispositivos como CLPs, IHMs, supervisórios, sensores e switches se comunicam.

Na área de **segurança**, isso é importante porque o sniffer ajuda a:
- identificar quem está se comunicando na rede
- verificar quais protocolos estão sendo usados
- detectar portas abertas e serviços expostos
- perceber tráfego incomum ou fora do padrão
- apoiar a investigação de falhas, ataques ou acessos indevidos

Em redes industriais, essa análise é especialmente útil porque muitos protocolos transportam comandos importantes para o processo. Observar o tráfego ajuda a entender o comportamento normal da rede e a perceber alterações suspeitas.

## Parâmetros úteis em um sniffer industrial

### 1. MAC
O endereço MAC identifica a interface de rede de um equipamento na camada de enlace. Ele é útil para descobrir quais dispositivos estão participando da comunicação local.

### 2. IP
O endereço IP identifica o host na rede. Com ele, é possível saber qual máquina está enviando ou recebendo dados.

### 3. Porta
A porta indica qual serviço ou aplicação está sendo usado. Em protocolos industriais, isso ajuda a reconhecer serviços como Modbus TCP, que normalmente usa a porta `502`.

### 4. EtherType
O EtherType informa qual protocolo está encapsulado no quadro Ethernet, como IPv4, ARP ou IPv6. Isso ajuda a entender a estrutura da comunicação.

### 5. Protocolo
Além de IP e porta, é importante identificar o protocolo industrial envolvido, como Modbus TCP, PROFINET ou EtherNet/IP.

### 6. Origem e destino
Observar origem e destino permite mapear quem iniciou a comunicação e para onde os dados estão indo.

### 7. Função ou tipo de operação
Quando o protocolo permite decodificação, é possível observar a operação executada, como leitura de registradores, escrita de bobinas ou resposta de erro.

### 8. Frequência e padrão do tráfego
A repetição e o intervalo entre pacotes ajudam a identificar o comportamento normal da rede e facilitam perceber anomalias.

## Relação com segurança

Na prática, um sniffer industrial não serve apenas para "ver pacotes". Ele também ajuda a responder perguntas importantes de segurança:
- Existe algum dispositivo desconhecido na rede?
- Alguém está acessando uma porta que normalmente não é usada?
- Há tráfego fora do padrão esperado para aquele processo?
- Um equipamento está enviando comandos que não deveria enviar?
- Ocorreu alguma mudança no comportamento da comunicação após uma falha ou incidente?

Por isso, o sniffer é uma ferramenta valiosa tanto para diagnóstico quanto para monitoramento e investigação.

## Resumo

O sniffer industrial permite analisar o tráfego da rede e compreender como os dispositivos se comunicam. Para segurança, ele é importante porque ajuda a identificar padrões, anomalias e possíveis riscos.

Os parâmetros mais úteis para começar são:
- MAC
- IP
- porta
- EtherType
- protocolo
- origem e destino
- função da mensagem
- frequência do tráfego
